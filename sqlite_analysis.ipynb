```python
# 🔹 Importando bibliotecas
import sqlite3
import pandas as pd
```

```python
# 🔹 Caminho do banco de dados SQLite
db_path = 'data/data.db'  # Certifique-se de colocar o caminho correto do banco
```

```python
# 🔹 Conectar ao banco e listar tabelas existentes
conn = sqlite3.connect(db_path)
query = "SELECT name FROM sqlite_master WHERE type='table';"
tables = pd.read_sql(query, conn)
conn.close()
tables
```

```python
# 🔹 Reabrindo conexão para inspecionar colunas da tabela
conn = sqlite3.connect(db_path)
query_columns = "PRAGMA table_info(dado_ventilador);"
columns_info = pd.read_sql(query_columns, conn)
conn.close()
columns_info
```

```python
# 🔹 Carregar os primeiros 10 registros da tabela
conn = sqlite3.connect(db_path)
query_data = "SELECT * FROM dado_ventilador LIMIT 10;"
df = pd.read_sql(query_data, conn)
conn.close()
df
```
